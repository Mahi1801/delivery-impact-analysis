```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Styling
plt.rcParams['figure.figsize'] = (12, 5)
plt.rcParams['font.size'] = 11
sns.set_style("whitegrid")
BEFORE_COLOR = "#4C72B0"
AFTER_COLOR  = "#DD8452"

df = pd.read_csv("../data/raw_orders.csv", parse_dates=["order_date", "order_time"])
print(df.shape)
df.head()
```

```python
print("=== Missing Values ===")
print(df.isnull().sum())

print("\n=== Dtypes ===")
print(df.dtypes)

print("\n=== Basic Stats ===")
df[['delivery_time_min', 'promised_eta_min', 'delivery_distance_km']].describe()
```

```python
# --- New Derived Columns ---

# Delay = how many minutes late or early
df['delay_min'] = df['delivery_time_min'] - df['promised_eta_min']

# On-time flag: 1 = delivered on time or early, 0 = late
df['on_time_flag'] = (df['delay_min'] <= 0).astype(int)

# SLA breach: more than 10 mins late
df['sla_breach'] = (df['delay_min'] > 10).astype(int)

# Cancelled flag
df['cancelled_flag'] = (df['order_status'] == 'Cancelled').astype(int)

# For cancelled orders: delivery_time is still recorded (dispatch time before cancel)
# Fill missing ratings with 0 for aggregation purposes
df['customer_rating_filled'] = df['customer_rating'].fillna(0)

print("New columns added:")
print(df[['delay_min', 'on_time_flag', 'sla_breach', 'cancelled_flag']].describe())
```

```python
# Remove extreme outliers in delivery time using IQR method
Q1 = df['delivery_time_min'].quantile(0.01)
Q3 = df['delivery_time_min'].quantile(0.99)
before_count = len(df)
df = df[(df['delivery_time_min'] >= Q1) & (df['delivery_time_min'] <= Q3)]
after_count = len(df)
print(f"Removed {before_count - after_count} outlier rows")
print(f"Clean dataset: {len(df):,} rows")
```

```python
df.to_csv("../data/clean_orders.csv", index=False)
print("✅ Saved: ../data/clean_orders.csv")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram
for period, color in [("Before", BEFORE_COLOR), ("After", AFTER_COLOR)]:
    subset = df[df['period'] == period]['delivery_time_min']
    axes[0].hist(subset, bins=40, alpha=0.6, color=color, label=period, edgecolor='white')

axes[0].set_title("Delivery Time Distribution: Before vs After", fontsize=13, fontweight='bold')
axes[0].set_xlabel("Delivery Time (mins)")
axes[0].set_ylabel("Number of Orders")
axes[0].legend()

# Box Plot
df_box = df[['period', 'delivery_time_min']]
sns.boxplot(data=df_box, x='period', y='delivery_time_min',
            palette={"Before": BEFORE_COLOR, "After": AFTER_COLOR},
            order=["Before", "After"], ax=axes[1])
axes[1].set_title("Delivery Time Spread: Before vs After", fontsize=13, fontweight='bold')
axes[1].set_xlabel("Period")
axes[1].set_ylabel("Delivery Time (mins)")

plt.tight_layout()
plt.savefig("../images/01_delivery_time_distribution.png", dpi=150, bbox_inches='tight')
plt.show()
```

```python
peak_kpi = df.groupby(['period', 'peak_flag']).agg(
    avg_delivery_time = ('delivery_time_min', 'mean'),
    on_time_pct       = ('on_time_flag', 'mean')
).round(2).reset_index()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Avg delivery time by period + peak
sns.barplot(data=peak_kpi, x='peak_flag', y='avg_delivery_time',
            hue='period', palette={"Before": BEFORE_COLOR, "After": AFTER_COLOR},
            ax=axes[0])
axes[0].set_title("Avg Delivery Time: Peak vs Non-Peak", fontsize=13, fontweight='bold')
axes[0].set_xlabel("Hour Type")
axes[0].set_ylabel("Avg Delivery Time (mins)")

# On-time % by period + peak
peak_kpi['on_time_pct_pct'] = peak_kpi['on_time_pct'] * 100
sns.barplot(data=peak_kpi, x='peak_flag', y='on_time_pct_pct',
            hue='period', palette={"Before": BEFORE_COLOR, "After": AFTER_COLOR},
            ax=axes[1])
axes[1].set_title("On-Time Delivery %: Peak vs Non-Peak", fontsize=13, fontweight='bold')
axes[1].set_xlabel("Hour Type")
axes[1].set_ylabel("On-Time Delivery %")
axes[1].set_ylim(0, 100)

plt.tight_layout()
plt.savefig("../images/02_peak_vs_nonpeak.png", dpi=150, bbox_inches='tight')
plt.show()
```

```python
kpi = df.groupby('period').agg(
    avg_delivery_time   = ('delivery_time_min', 'mean'),
    p90_delivery_time   = ('delivery_time_min', lambda x: x.quantile(0.9)),
    on_time_pct         = ('on_time_flag', 'mean'),
    cancellation_rate   = ('cancelled_flag', 'mean'),
    sla_breach_rate     = ('sla_breach', 'mean'),
    avg_rating          = ('customer_rating', 'mean'),
    total_orders        = ('order_id', 'count')
).round(3)

kpi['on_time_pct']       = (kpi['on_time_pct'] * 100).round(1)
kpi['cancellation_rate'] = (kpi['cancellation_rate'] * 100).round(1)
kpi['sla_breach_rate']   = (kpi['sla_breach_rate'] * 100).round(1)

kpi_display = kpi.loc[['Before', 'After']].copy()
kpi_display.columns = ['Avg Del. Time', 'P90 Del. Time', 'On-Time %',
                        'Cancel Rate %', 'SLA Breach %', 'Avg Rating', 'Total Orders']

# Calculate % change
change = ((kpi_display.loc['After'] - kpi_display.loc['Before']) / kpi_display.loc['Before'] * 100).round(1)
kpi_display.loc['% Change'] = change.astype(str) + '%'

print("=== KPI SUMMARY ===")
print(kpi_display.to_string())
```

```python
zone_kpi = df.groupby(['period', 'zone']).agg(
    avg_delivery_time = ('delivery_time_min', 'mean'),
    cancellation_rate = ('cancelled_flag', 'mean')
).round(3).reset_index()

zone_kpi['cancellation_pct'] = (zone_kpi['cancellation_rate'] * 100).round(1)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.barplot(data=zone_kpi, x='zone', y='avg_delivery_time',
            hue='period', palette={"Before": BEFORE_COLOR, "After": AFTER_COLOR},
            order=['High','Medium','Low'], ax=axes[0])
axes[0].set_title("Avg Delivery Time by Zone", fontsize=13, fontweight='bold')

sns.barplot(data=zone_kpi, x='zone', y='cancellation_pct',
            hue='period', palette={"Before": BEFORE_COLOR, "After": AFTER_COLOR},
            order=['High','Medium','Low'], ax=axes[1])
axes[1].set_title("Cancellation Rate (%) by Zone", fontsize=13, fontweight='bold')

plt.tight_layout()
plt.savefig("../images/03_zone_impact.png", dpi=150, bbox_inches='tight')
plt.show()
```

```python
dist_kpi = df.groupby(['period', 'distance_bucket']).agg(
    avg_delivery_time = ('delivery_time_min', 'mean'),
    sla_breach_pct    = ('sla_breach', 'mean')
).round(3).reset_index()

order = ['Short (0-3 km)', 'Medium (3-7 km)', 'Long (7+ km)']

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.barplot(data=dist_kpi, x='distance_bucket', y='avg_delivery_time',
            hue='period', palette={"Before": BEFORE_COLOR, "After": AFTER_COLOR},
            order=order, ax=axes[0])
axes[0].set_title("Avg Delivery Time by Distance", fontsize=13, fontweight='bold')
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation=15)

dist_kpi['sla_pct'] = dist_kpi['sla_breach_pct'] * 100
sns.barplot(data=dist_kpi, x='distance_bucket', y='sla_pct',
            hue='period', palette={"Before": BEFORE_COLOR, "After": AFTER_COLOR},
            order=order, ax=axes[1])
axes[1].set_title("SLA Breach Rate (%) by Distance", fontsize=13, fontweight='bold')
axes[1].set_xticklabels(axes[1].get_xticklabels(), rotation=15)

plt.tight_layout()
plt.savefig("../images/04_distance_impact.png", dpi=150, bbox_inches='tight')
plt.show()
```

```python
daily = df.groupby(['order_date', 'period']).agg(
    avg_delivery_time = ('delivery_time_min', 'mean')
).reset_index()

plt.figure(figsize=(14, 5))
for period, color in [("Before", BEFORE_COLOR), ("After", AFTER_COLOR)]:
    subset = daily[daily['period'] == period]
    plt.plot(subset['order_date'], subset['avg_delivery_time'],
             color=color, linewidth=2, label=period, alpha=0.8)

plt.axvline(x=daily[daily['period']=='After']['order_date'].min(),
            color='red', linestyle='--', linewidth=1.5, label='Feature Removed')

plt.title("Daily Avg Delivery Time — Before vs After Feature Removal",
          fontsize=13, fontweight='bold')
plt.xlabel("Date")
plt.ylabel("Avg Delivery Time (mins)")
plt.legend()
plt.tight_layout()
plt.savefig("../images/05_daily_trend.png", dpi=150, bbox_inches='tight')
plt.show()
```
