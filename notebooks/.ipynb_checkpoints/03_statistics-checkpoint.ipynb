```python
import pandas as pd
import numpy as np
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv("../data/clean_orders.csv")
before = df[df['period'] == 'Before']
after  = df[df['period'] == 'After']

print(f"Before: {len(before):,} orders | After: {len(after):,} orders")
```

```python
t_stat, p_value = stats.ttest_ind(
    before['delivery_time_min'],
    after['delivery_time_min'],
    equal_var=False  # Welch's t-test (safer assumption)
)

mean_before = before['delivery_time_min'].mean()
mean_after  = after['delivery_time_min'].mean()
pct_change  = ((mean_after - mean_before) / mean_before) * 100

print("=" * 55)
print("  T-TEST: Average Delivery Time")
print("=" * 55)
print(f"  Mean (Before): {mean_before:.2f} mins")
print(f"  Mean (After):  {mean_after:.2f} mins")
print(f"  Change:        {pct_change:+.1f}%")
print(f"  T-statistic:   {t_stat:.4f}")
print(f"  P-value:       {p_value:.6f}")
print(f"  Significant?   {'✅ YES (p < 0.05)' if p_value < 0.05 else '❌ NO'}")
```

```python
def cohens_d(group1, group2):
    """Cohen's d effect size."""
    n1, n2   = len(group1), len(group2)
    var1     = group1.var(ddof=1)
    var2     = group2.var(ddof=1)
    pooled_std = np.sqrt(((n1-1)*var1 + (n2-1)*var2) / (n1+n2-2))
    return (group2.mean() - group1.mean()) / pooled_std

d = cohens_d(before['delivery_time_min'], after['delivery_time_min'])

magnitude = (
    "Negligible" if abs(d) < 0.2 else
    "Small"      if abs(d) < 0.5 else
    "Medium"     if abs(d) < 0.8 else
    "Large"
)

print("=" * 55)
print("  EFFECT SIZE: Cohen's d")
print("=" * 55)
print(f"  Cohen's d:  {d:.4f}")
print(f"  Magnitude:  {magnitude}")
print("  (Small=0.2, Medium=0.5, Large=0.8)")
```

```python
# Contingency table: On-Time vs Late, Before vs After
ct = pd.crosstab(df['period'], df['on_time_flag'])
ct.columns = ['Late', 'On-Time']
print("\nContingency Table:")
print(ct)

chi2, p_chi, dof, expected = stats.chi2_contingency(ct)

on_time_before = before['on_time_flag'].mean() * 100
on_time_after  = after['on_time_flag'].mean() * 100

print("\n" + "=" * 55)
print("  CHI-SQUARE TEST: On-Time Delivery %")
print("=" * 55)
print(f"  On-Time % (Before): {on_time_before:.1f}%")
print(f"  On-Time % (After):  {on_time_after:.1f}%")
print(f"  Chi2 statistic:     {chi2:.4f}")
print(f"  P-value:            {p_chi:.6f}")
print(f"  Significant?        {'✅ YES (p < 0.05)' if p_chi < 0.05 else '❌ NO'}")
```

```python
ct_cancel = pd.crosstab(df['period'], df['cancelled_flag'])
ct_cancel.columns = ['Delivered', 'Cancelled']

chi2_c, p_c, _, _ = stats.chi2_contingency(ct_cancel)

cancel_before = before['cancelled_flag'].mean() * 100
cancel_after  = after['cancelled_flag'].mean() * 100

print("=" * 55)
print("  CHI-SQUARE TEST: Cancellation Rate")
print("=" * 55)
print(f"  Cancel Rate (Before): {cancel_before:.2f}%")
print(f"  Cancel Rate (After):  {cancel_after:.2f}%")
print(f"  P-value:              {p_c:.6f}")
print(f"  Significant?          {'✅ YES (p < 0.05)' if p_c < 0.05 else '❌ NO'}")
```

```python
print("=" * 55)
print("  PEAK HOUR IMPACT ANALYSIS")
print("=" * 55)

for peak in ['Peak', 'Non-Peak']:
    b = before[before['peak_flag'] == peak]['delivery_time_min']
    a = after[after['peak_flag'] == peak]['delivery_time_min']
    t, p = stats.ttest_ind(b, a, equal_var=False)
    pct  = ((a.mean() - b.mean()) / b.mean()) * 100
    print(f"\n  [{peak}]")
    print(f"    Before: {b.mean():.2f} min | After: {a.mean():.2f} min")
    print(f"    Change: {pct:+.1f}% | p-value: {p:.6f}")
```
